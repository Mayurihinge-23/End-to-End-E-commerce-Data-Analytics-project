```python
# libraries import
import pandas as pd
import numpy as np
```

```python
# import dataset
df=pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# check missing value
df.isnull().sum()
```

```python
# check duplicate record
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
# check duplicate record
df.duplicated().sum()
```

```python
#replace invalid values
# replace common invalid values with NaN
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

```python
# remove extra space
df=df.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

```python
# standardrize text format
# convert text into proper text

df['Customer_Name']=df['Customer_Name'].str.title()
df['City']=df['City'].str.title()
df['State']=df['State'].str.title()
```

```python
# validate email address
# keep only valid email record
df=df[df['Email'].str.contains('@',na=False)]
```

```python
# convert date column
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# convert numeric column
cols=['Qty','Unit_Price','Discount']
for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
# remove invalid quantity
# remove negative and zero quantity

df=df[df['Qty']>0]
```

```python
df = df.copy()
```

```python
# filling missing value
df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('Unknown')
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature Engineering
# TotalSales

df['Total_sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
# Net_Amount
df['Net_Amount']=df['Total_sales']-(df['Total_sales']*df['Discount']/100)
```

```python
df
```

```python
df['Profit']=df['Net_Amount']*0.20
```

```python
df
```

```python
# Order Month
df['Month']=df['Order_Date'].dt.month_name()
```

```python
df
```

```python
# order year
df['Year']=df['Order_Date'].dt.year
```

```python
df
```

```python
# weekday
df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# final data visualization
df.info()
```
